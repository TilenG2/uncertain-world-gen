```python
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from math import pi, cos, sin
def get_ellipse(x, y, a, b, rot_rad):
    # x=5       #x-position of the center
    # y=5      #y-position of the center
    # a=6       #radius on the x-axis
    # b=4      #radius on the y-axis
    # rot_rad=5*pi/6 #rotation angle

    t = np.linspace(0, 2*pi, 100)
    angles = np.random.uniform(0,2*pi, 1000)
    muls = np.abs(np.random.normal(0, 0.4, 1000))
    Ell = []
    for angle, mul in zip(angles, muls):
        Ell.append([mul*a*np.cos(angle), mul*b*np.sin(angle)])
    Ell = np.array(Ell).T #u,v removed to keep the same center location
    R_rot = np.array([[cos(rot_rad) , -sin(rot_rad)],[sin(rot_rad) , cos(rot_rad)]])  #2-D rotation matrix

    Ell_rot = np.zeros((2,Ell.shape[1]))
    for i in range(Ell.shape[1]):
        Ell_rot[:,i] = np.dot(R_rot,Ell[:,i])
    Ell_rot[0,:] += x
    Ell_rot[1,:] += y
    return Ell_rot
```

```python
import numpy as np
from matplotlib import pyplot as plt

Ell1= get_ellipse(5,5,6,3,9*pi/12)
Ell2= get_ellipse(2.5,2.5,6,3,9*pi/12)

plt.scatter( Ell2[0,:] , Ell2[1,:])     #initial ellipse
plt.scatter( Ell1[0,:] , Ell1[1,:])    #rotated ellipse
plt.grid(color='lightgray',linestyle='--')
plt.xlim([-5,15])
plt.ylim([-5,14])
plt.show()
```
